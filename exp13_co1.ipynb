```python
import numpy as np

# Actual probability distribution
actual = np.array([0.50, 0.30, 0.20])

# Predicted probability distribution
predicted = np.array([0.40, 0.40, 0.20])

# KL divergence: D(actual || predicted)
kl_divergence = np.sum(actual * np.log(actual / predicted))

print(f"KL divergence: {kl_divergence:.4f} nats")

if np.isclose(kl_divergence, 0):
    print("The distributions match; no information is lost.")
else:
    print("A non-zero value indicates information loss when using "
          "the predicted distribution instead of the actual distribution.")
```
