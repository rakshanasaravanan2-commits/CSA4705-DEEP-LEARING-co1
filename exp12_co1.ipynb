```python
import numpy as np

# Actual labels: 1 = spam, 0 = not spam
actual = np.array([1, 0, 1, 0, 1])

# Model's predicted probability that each email is spam
predicted = np.array([0.90, 0.20, 0.70, 0.60, 0.40])

# Avoid log(0)
epsilon = 1e-15
predicted = np.clip(predicted, epsilon, 1 - epsilon)

# Binary cross-entropy for each email and average loss
losses = -(
    actual * np.log(predicted)
    + (1 - actual) * np.log(1 - predicted)
)
average_loss = np.mean(losses)

print("Cross-entropy loss per email:", np.round(losses, 4))
print(f"Average cross-entropy loss: {average_loss:.4f}")

for i, loss in enumerate(losses, start=1):
    print(f"Email {i}: loss = {loss:.4f}")
```
