```python
from dataclasses import dataclass

@dataclass
class Vector3D:
    x: float
    y: float
    z: float

    def __add__(self, other):
        return Vector3D(self.x + other.x, self.y + other.y, self.z + other.z)

    def __sub__(self, other):
        return Vector3D(self.x - other.x, self.y - other.y, self.z - other.z)

    def __mul__(self, scalar):
        return Vector3D(self.x * scalar, self.y * scalar, self.z * scalar)

    def __rmul__(self, scalar):
        return self * scalar

    def __str__(self):
        return f"({self.x}, {self.y}, {self.z})"


# Robot's current position
position = Vector3D(2, 3, 1)

# Direction of movement, scaled by travel distance
direction = Vector3D(1, 0, 2)
distance = 4
movement = direction * distance

# Calculate the new position
new_position = position + movement

print("Current position:", position)
print("Movement vector:", movement)
print("New position:", new_position)

# Example subtraction: calculate the vector from the origin
origin = Vector3D(0, 0, 0)
relative_position = new_position - origin
print("Position relative to origin:", relative_position)
```
