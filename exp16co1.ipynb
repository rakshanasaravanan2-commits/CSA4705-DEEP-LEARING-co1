```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Features: area (sq ft), bedrooms, distance from city centre (km)
X = np.array([
    [1200, 2, 10],
    [1500, 3, 8],
    [1800, 3, 6],
    [2200, 4, 5],
    [2500, 4, 3],
    [3000, 5, 2]
])

# House prices in thousands of dollars
y = np.array([200, 250, 300, 360, 410, 500])

model = LinearRegression()
model.fit(X, y)

predictions = model.predict(X)

print("Predicted prices (thousands of dollars):")
print(np.round(predictions, 2))

print(f"\nMean Absolute Error: {mean_absolute_error(y, predictions):.2f}")
print(f"Mean Squared Error: {mean_squared_error(y, predictions):.2f}")

# Predict a 2,000 sq ft, 3-bedroom house 4 km from the city centre
new_house = np.array([[2000, 3, 4]])
predicted_price = model.predict(new_house)[0]
print(f"\nPredicted price for new house: ${predicted_price:.2f} thousand")
```
