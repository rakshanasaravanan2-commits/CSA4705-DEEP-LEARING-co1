```python
import math
from collections import Counter

# Loan applicant risk labels
applicants = [
    "Low-risk", "Low-risk", "Low-risk", "Low-risk",
    "High-risk", "High-risk"
]

def calculate_entropy(labels):
    counts = Counter(labels)
    total = len(labels)

    entropy = 0
    for count in counts.values():
        probability = count / total
        entropy -= probability * math.log2(probability)

    return entropy

entropy = calculate_entropy(applicants)

print(f"Applicant counts: {Counter(applicants)}")
print(f"Entropy: {entropy:.4f} bits")

if entropy == 0:
    print("Classification uncertainty: None")
else:
    print("Classification uncertainty: Present")
```
