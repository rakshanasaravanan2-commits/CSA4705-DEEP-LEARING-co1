```python
# Cost function: f(x) = (x - 5)^2 + 10
# Its minimum is at x = 5

def cost(x):
    return (x - 5) ** 2 + 10

def gradient(x):
    return 2 * (x - 5)

x = 0.0
learning_rate = 0.1
iterations = 50

for _ in range(iterations):
    x -= learning_rate * gradient(x)

print(f"Minimum-cost parameter: {x:.4f}")
print(f"Minimum production cost: {cost(x):.4f}")
```
