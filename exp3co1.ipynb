```python
import numpy as np

matrix = np.array([
    [3, 2],
    [5, 3]
], dtype=float)

determinant = np.linalg.det(matrix)
rank = np.linalg.matrix_rank(matrix)

print("Matrix:")
print(matrix)
print("Determinant:", determinant)
print("Rank:", rank)

if np.isclose(determinant, 0):
    print("Inverse: does not exist")
    print("Suitable for reversible encryption: No")
else:
    inverse = np.linalg.inv(matrix)
    print("Inverse:")
    print(inverse)
    print("Suitable for reversible encryption: Yes")
```
