```python
import numpy as np

# A 2D rotation matrix: rotates coordinates by 90 degrees
rotation = np.array([
    [0, -1],
    [1,  0]
])

# A scaling matrix: doubles x and triples y
scaling = np.array([
    [2, 0],
    [0, 3]
])

# Matrix multiplication: apply scaling, then rotation
combined = rotation @ scaling

# Transpose the combined matrix
transposed = combined.T

# A sample image coordinate
point = np.array([1, 2])

# Transform the point
transformed_point = combined @ point

print("Rotation matrix:\n", rotation)
print("Scaling matrix:\n", scaling)
print("\nCombined matrix (rotation @ scaling):\n", combined)
print("\nTransposed combined matrix:\n", transposed)
print("\nOriginal point:", point)
print("Transformed point:", transformed_point)

# Compare with the reverse multiplication order
reverse_order = scaling @ rotation
print("\nScaling @ rotation:\n", reverse_order)
print("\nDo the multiplication orders match?",
      np.array_equal(combined, reverse_order))
```
