```python
# P(Machine A) and P(Machine B)
p_machine_a = 0.60
p_machine_b = 0.40

# Defect rates
p_defect_given_a = 0.02
p_defect_given_b = 0.05

# Total probability of a defective product
p_defective = (
    p_defect_given_a * p_machine_a
    + p_defect_given_b * p_machine_b
)

# Bayes' theorem: probability defective product came from Machine A
p_a_given_defective = (
    p_defect_given_a * p_machine_a
    / p_defective
)

# Probability it came from Machine B
p_b_given_defective = (
    p_defect_given_b * p_machine_b
    / p_defective
)

print(f"P(Machine A | Defective): {p_a_given_defective:.2%}")
print(f"P(Machine B | Defective): {p_b_given_defective:.2%}")
```
