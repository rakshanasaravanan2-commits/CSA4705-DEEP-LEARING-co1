```python
import numpy as np

# Example dataset: y = 2x + 1
X = np.array([1, 2, 3, 4, 5], dtype=float)
y = 2 * X + 1

learning_rate = 0.01
epochs = 100
batch_size = 2

def predict(parameters, x):
    return parameters[0] * x + parameters[1]

def gradient(parameters, x_batch, y_batch):
    errors = predict(parameters, x_batch) - y_batch
    grad_slope = 2 * np.mean(errors * x_batch)
    grad_intercept = 2 * np.mean(errors)
    return np.array([grad_slope, grad_intercept])

def loss(parameters):
    return np.mean((predict(parameters, X) - y) ** 2)

def train(method):
    parameters = np.array([0.0, 0.0])
    losses = []
    updates = 0

    for _ in range(epochs):
        if method == "batch":
            parameters -= learning_rate * gradient(parameters, X, y)
            updates += 1

        elif method == "stochastic":
            for i in np.random.permutation(len(X)):
                parameters -= learning_rate * gradient(
                    parameters, X[i:i + 1], y[i:i + 1]
                )
                updates += 1

        elif method == "mini-batch":
            indices = np.random.permutation(len(X))
            for start in range(0, len(X), batch_size):
                batch_indices = indices[start:start + batch_size]
                parameters -= learning_rate * gradient(
                    parameters, X[batch_indices], y[batch_indices]
                )
                updates += 1

        losses.append(loss(parameters))

    return parameters, losses, updates

np.random.seed(42)

for method in ["batch", "stochastic", "mini-batch"]:
    parameters, losses, updates = train(method)
    print(f"{method.title()} gradient descent:")
    print(f"  Parameters: slope={parameters[0]:.4f}, "
          f"intercept={parameters[1]:.4f}")
    print(f"  Final loss: {losses[-1]:.6f}")
    print(f"  Parameter updates: {updates}")
```
