```python
# Example values
prevalence = 0.01     # P(Disease)
sensitivity = 0.95    # P(Positive | Disease)
specificity = 0.90    # P(Negative | No Disease)

# P(Positive | No Disease)
false_positive_rate = 1 - specificity

# Bayes' theorem
p_positive = (
    sensitivity * prevalence
    + false_positive_rate * (1 - prevalence)
)

p_disease_given_positive = (
    sensitivity * prevalence
    / p_positive
)

print(f"Probability patient has disease after positive test: "
      f"{p_disease_given_positive:.2%}")
```
