```python
import numpy as np

marks = np.array([78, 85, 92, 68, 88, 75, 95, 81])

mean = np.mean(marks)
variance = np.var(marks)           # Population variance
standard_deviation = np.std(marks)  # Population standard deviation

print("Marks:", marks)
print(f"Mean: {mean:.2f}")
print(f"Variance: {variance:.2f}")

print(f"Standard deviation: {standard_deviation:.2f}")

if standard_deviation < 10:
    print("Interpretation: Marks are relatively consistent.")
else:
    print("Interpretation: Marks show considerable variation.")
```
