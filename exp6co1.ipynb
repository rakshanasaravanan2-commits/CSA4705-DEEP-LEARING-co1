```python
import numpy as np

customers = {
    "Customer A": np.array([5, 2, 1]),
    "Customer B": np.array([4, 3, 1]),
    "Customer C": np.array([1, 5, 4])
}

def euclidean_distance(a, b):
    return np.sqrt(np.sum((a - b) ** 2))

def manhattan_distance(a, b):
    return np.sum(np.abs(a - b))

names = list(customers)

for i in range(len(names)):
    for j in range(i + 1, len(names)):
        name_a, name_b = names[i], names[j]
        a, b = customers[name_a], customers[name_b]

        euclidean = euclidean_distance(a, b)
        manhattan = manhattan_distance(a, b)

        print(f"{name_a} vs {name_b}")
        print(f"  Euclidean distance: {euclidean:.2f}")
        print(f"  Manhattan distance: {manhattan:.2f}")

# Find the closest customer pair using each metric
pairs = [
    (names[i], names[j], euclidean_distance(customers[names[i]], customers[names[j]]),
     manhattan_distance(customers[names[i]], customers[names[j]]))
    for i in range(len(names))
    for j in range(i + 1, len(names))
]

closest_euclidean = min(pairs, key=lambda pair: pair[2])
closest_manhattan = min(pairs, key=lambda pair: pair[3])

print("\nClosest pair by Euclidean distance:",
      closest_euclidean[0], "and", closest_euclidean[1])
print("Closest pair by Manhattan distance:",
      closest_manhattan[0], "and", closest_manhattan[1])
```
