```python
import numpy as np
from scipy import stats

# Generate temperature observations expected to follow a normal distribution
np.random.seed(42)
expected_mean = 25
expected_std = 5
temperatures = np.random.normal(expected_mean, expected_std, 1000)

# Estimate distribution parameters from the observations
estimated_mean = np.mean(temperatures)
estimated_std = np.std(temperatures, ddof=1)

# Test whether observations follow the expected normal distribution
ks_statistic, p_value = stats.kstest(
    temperatures,
    "norm",
    args=(expected_mean, expected_std)
)

print(f"Estimated mean: {estimated_mean:.2f} °C")
print(f"Estimated standard deviation: {estimated_std:.2f} °C")
print(f"KS statistic: {ks_statistic:.4f}")
print(f"P-value: {p_value:.4f}")

if p_value > 0.05:
    print("The observations are consistent with the expected distribution.")
else:
    print("The observations differ significantly from the expected distribution.")
```
